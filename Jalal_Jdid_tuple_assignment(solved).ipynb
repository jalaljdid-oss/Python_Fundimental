```python
'''Create a tuple with the first 10 positive integers. Print the tuple.
'''
tup=tuple([i for i in range(1,11)])
print(tup)
```

```python
'''Print the first, middle, and last elements of the tuple created in Assignment 1.
'''
first,*middle,last=tup
print(f"first is {first}\nthe middle elements is {middle}\nlast is {last}")
```

```python
'''Print the first three elements, the last three elements, and the elements from index 2 to 5 of the tuple created in Assignment 1.
'''
print(f"the First 3 element is {tup[0:3]}\nthe element from 2 to 5 is {tup[2:6]}\nthe Last element is {tup[len(tup)-3:]}")
```

```python
'''Create a nested tuple representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column.
'''
import random
nest_tup=tuple([(random.randint(1,100),random.randint(1,100),random.randint(1,100)) for i in range(3) ])
print (nest_tup)
for row in nest_tup:
    print(row)
print("the second Row and Third element is ",nest_tup[1][2])
```

```python
'''Concatenate two tuples: (1, 2, 3) and (4, 5, 6). Print the resulting tuple.
'''
tup1=(1,2,3)
tup2=(4,5,6)
con_tup=tup1+tup2
print(f"concatination ot two tuples is mit + sign is {con_tup}")
```

```python
'''Create a tuple with duplicate elements and count the occurrences of an element. Find the index of the first occurrence of an element in the tuple.
'''
tup3=tuple([1,1,1,1,1,3,3,3,3,3,6,6,6,6,6])
print(f"the count of element 3 is {tup3.count(3)}")
```

```python
'''Create a tuple with 5 elements and unpack it into 5 variables. Print the variables.
'''
pac_tup=tuple([3,2,5,9,True])
a,b,c,d,e=pac_tup
print(f"a , b, c, d, e = {a} ,{b} ,{c} ,{d} ,{e}")
```

```python
'''Convert a list of the first 5 positive integers to a tuple. Print the tuple.
'''
li=[i for i in range(1,6)]
lis_to_tupl=tuple(li)

print(lis_to_tupl)
```

```python
'''Create a tuple containing 3 tuples, each with 3 elements. Print the tuple of tuples.
'''
import random
nested_tup=((random.randint(1,100),random.randint(1,100),random.randint(1,100)) for i in range(3))

for item in nested_tup:
    print(item)
```

```python
'''Create a tuple with the first 5 positive integers. Convert it to a list, append the number 6, and convert it back to a tuple. Print the resulting tuple.
'''
tup3=(1,2,3,4,5)
tup_to_li=list(tup3)
tup_to_li.append(6)
tup3=tuple(tup_to_li)
print(tup3)
```

```python
'''Create a tuple with the characters of a string. Join the tuple elements into a single string. Print the string.'''

tup4=('jalal','mohammad','jdid')
s=""
for each in tup4:
    s+=f"{each} "

print (s)
```

```python
'''Create a dictionary with tuple keys and integer values. Print the dictionary.
'''
dic={(1,2):8,(2,3):100,(25,3):55
    
}
print(dic)
for key,value in dic.items():
    print(f"{key}:{value}")
```

```python
'''Create a nested tuple and iterate over the elements, printing each element.
'''
tpl=tuple([(i,j,i+j) for i in range(5,10) for j in range(8,10)])
print(tpl)
for item in tpl:
   print(item)
```

```python
'''Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.
'''
import random
def min_max_sum_elements(li):
    return min(li),max(li),sum(li)


tp=tuple([random.randint(1,50) for i in range(10)])
print(tp)
mn,mx,su=min_max_sum_elements(tp)
print(f"max is {mx}\nmin is {mn}\nsum is {su}")
        

```
