```python
'''### Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

'''
def transpose(matrix):
    result=[[0,0,0],[0,0,0],[0,0,0]] 
    
    #result.clear()
    
   
    for r in range(len(matrix)):
        for c in range(len(matrix)):
            result[c][r]=matrix[r][c]
            
    
    return result

 
mx=[[1,2,3],
    [4,5,6],
    [7,8,9]]


print("your matrixx is:")
for row in mx:
    print(row)
result=transpose(mx)
print("your transposed_matrixx is:")
for row in result:
    print(row)

```

```python
'''
### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.


'''
def flatten(nested_list):
    result=[]
    for item in nested_list:  
        if  isinstance(item,list):
            result.extend(flatten(item)) # recursive function
        else:
            result.append(item)
    return result


nested_li=[[1,2,3],
    [4,5,6],
    [7,8],5,[[1],8]]

result=flatten(nested_li)
print(result)

```

```python
'''### Assignment 12: List Zipping

Create two lists of the same length. Use the `zip` function to combine these lists into a list of tuples and print the result.

'''

names=['jalal','ali','fadee']
scores=[66,55,73]
result=[]
for n,s in zip(names,scores):
    result.append((n,s))
print(f"list of tuples {result}")
```

```python
'''
### Assignment 13: List Reversal

Write a function that takes a list and returns a new list with the elements in reverse order. Print the original and reversed lists.

'''
def reversed_list(li):
   return li[::-1]

li_4=[5,3,6,9,8,7,4,1]

print(f" the reversed List is {reversed_list(li_4)}")
```

```python
'''### Assignment 14: List Rotation

Write a function that rotates a list by n positions. Print the original and rotated lists.

'''
#position form left to right count
def rotate_list(li,n):
    if n>len(li):
        real_n=n%len(li)
    return li[real_n:]+li[:real_n]

li_5=[1,2,3,4,5,6,7]
n=int(input("number to rotate the list:"))

print(f"rotated list by {n} is {rotate_list(li_5,n)}")
```

```python
'''
### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.
'''
def intersected_listed(li_1,li_2):
   s1=set(li_1)
   s2=set(li_2)
   return s1.intersection(s2) 

li1=[1,2,3,4,5,5,5,5,5,7,7,7,10,31]
li2=[3,4,5,3,3,31,1]

print(intersected_listed(li1,li2))
```
