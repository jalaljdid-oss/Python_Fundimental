```python
'''Iterate over the dictionary created in Assignment 1 and print each key-value pair.'''
for key,value in my_square_dict.items():
    print(f"{key}:{value}")
```

```python
'''Create a new dictionary containing the cubes of the first 10 positive integers using a dictionary comprehension. Print the new dictionary.'''
my_cube_dict={i:i**3 for i in range(1,11)}
print(my_cube_dict)
```

```python
'''Create two dictionaries: one with keys as the first 5 positive integers and values as their squares, 
and another with keys as the next 5 positive integers and values as their squares. Merge these dictionaries into a single dictionary and print it.
'''
firstfive_dic={i:i**2 for i in range(1,6)}
print(firstfive_dic)
nextfive_dic={i:i**2 for i in range(6,11)}
print(nextfive_dic)

my_merged_dic={**firstfive_dic,**nextfive_dic}
print(my_merged_dic)
```

```python
'''Create a dictionary where the keys are the first 5 positive integers and the values are lists containing the first 5 multiples of the key. Print the dictionary.'''

dic_list={i:[i*j for j in range(1,6)] for i in range(1,6) }
print(dic_list)
```

```python
'''Create a dictionary where the keys are the first 5 positive integers and the values are tuples containing the key and its square. Print the dictionary.'''

dic_tup={i:(i,i**2) for i in range(1,6)}
print(dic_tup)
```

```python
'''Create a dictionary with the first 5 positive integers as keys and their squares as values. Convert the dictionary to a list of tuples and print it.'''
dic_5_squares={i:i**2 for i in range(1,6)}
print(dic_5_squares)

list_tup=list(tuple(dic_5_squares.items()))
print(list_tup)
```

```python
'''Create a dictionary with the first 10 positive integers as keys and their squares as values. Create a new dictionary containing only the key-value pairs where the key is even. 
Print the new dictionary.'''
dic_10_squares={i:i**2 for i in range(1,11)}
print(dic_10_squares)

dic_10_even_key_squares={i:i**2 for i in range(1,11) if i%2==0}
print(dic_10_even_key_squares)
```

```python
'''Write a function that takes a string and returns a dictionary with the count of each character in the string. Print the dictionary.'''

def count_char_of_a_string(s="test"):
    
    my_dic={}
    
    for ch in s:
        my_dic[ch]=my_dic.get(ch,0)+1
        
    return my_dic
               
    
count_char_of_a_string("jalal  kkkk jdid")
```
