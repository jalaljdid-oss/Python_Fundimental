```python
'''
Assignment 14: Prime Number Check
Write a program that checks if a number input by the user is a prime number
using a for loop.
'''

while True:
  try:
    num = int(input("Enter a number: "))
    break
  except ValueError:
    print("Invalid input. Please enter an integer.")

if num <= 1:
  print(f"{num} is not a prime number.")
elif num == 2:
  print(f"{num} is a prime number.")
else:
  is_prime = True
  for i in range(2, num):
    if (num % i) == 0:
      is_prime = False
      break
  if is_prime:
    print(f"{num} is a prime number.")
  else:
    print(f"{num} is not a prime number.")
```

```python
'''
Assignment 15: Fibonacci Sequence
Write a program that prints the first n Fibonacci numbers, where n is input by the user.
'''
while True:
  try:
    num = int(input("Enter a number to print the first n Fibonacci numbers: "))
    break
  except ValueError:
    print("Invalid input. Please enter an integer.")

if num < 0:
  print("number is negative.")
elif num == 0:
  print("The Fibonacci sequence for 0 is: []")
elif num == 1:
  print("The Fibonacci sequence for 1 is: [0]")
else:
  fib = [0,1]
  while len(fib) < num:
    fib.append(fib[len(fib)-1]+fib[len(fib)-2])

  print(f"The first {num} Fibonacci numbers are: {fib}")
```
