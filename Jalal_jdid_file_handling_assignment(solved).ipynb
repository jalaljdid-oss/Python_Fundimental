```python
'''Write a function that reads the contents of a file named `sample.txt` and prints each line.'''
import os
def readfile(filepath):
   if os.path.exists(filepath) : 
    with open(filepath,"r") as file:
        content= file.read()
    print (content)    
    
#main
#print(os.getcwd())

file_name="sample.txt"
readfile(file_name)
```

```python
'''Write a function that writes a list of strings to a file named `output.txt`, with each string on a new line.'''
import os
def writ_to_file(filepath,li):
   if os.path.exists(filepath) :
    with open(filepath,'w') as file:
        for item in li:
            file.write(item+"\n")

#main

fileName="output.txt"            
li=["hello I am jalal","hello I am Joud","hello I am zakrya"]
writ_to_file(fileName,li)

#check if the code working well call last function readfile()
readfile(fileName)
```

```python
'''Write a function that copies the contents of a file named `source.txt` to a new file named `destination.txt`.'''
import os
def copyfile(source,dist):
   if os.path.exists(source) : 
    with open(source,'r') as sf:
        tobecopied=sf.read()
    with open(dist,'w') as wf:
        wf.write(tobecopied)
        
#main

file1="source.txt"
file2="destination.txt"

copyfile(file1,file2)

# check call readfile() method

readfile(file2)
```

```python
'''Write a function that appends a given string to the end of a file named `log.txt`.'''
import os
def appendStrTofile(filepath,str):
   if os.path.exists(filepath) : 
    with open(filepath,'a') as file:
        file.write(str)
        
#main
fileName="log.txt"
str="hello jalal"
appendStrTofile(fileName,str)
readfile(fileName)
```

```python
'''write a function that reads the contents of a file named `document.txt` and returns the number of words in the file.
'''
import os
def countFileWords(filepath):
   if os.path.exists(filepath) : 
    with open(filepath,'r') as file:
        return sum(len(line.split()) for line in file.readlines())
    
#main 
fileName="sample.txt"
print(f"wordscount = {countFileWords(fileName)}")
```

```python
'''Write a function that finds and replaces all occurrences of a given word in a file named `data.txt` with another word.'''
import os
def replaceWordInFile(filepath,old_string,new_string):
   if os.path.exists(filepath) :  
    with open("temp123.txt",'w') as newfile: # create temporal file
        with open(filepath,'r') as orginal:
            for num,line in enumerate(orginal): #to reduce using memory lin by line read
                newline=line.replace(old_string,new_string)
                newfile.write(newline)
    # deleting orginal file
    if os.path.exists(filepath):
        os.remove(filepath)
    #renaming the new file as same file orginal file name
    if not os.path.exists(filepath):
        os.rename("temp123.txt",filepath)
        
#main

fileName="data.txt"
print(f"the file before updated")
readfile(fileName)
replaceWordInFile(fileName,'joud','jalal')
print(f"the file AFTER updated\n")
readfile(fileName)
           
    
            
    
```

```python
'''Write a function that reads the contents of a file named `stats.txt` and returns the number of lines, words, and characters in the file.'''

import os
def countFile(filepath):
   if os.path.exists(filepath) : 
    with open(filepath,'r') as file:
        content=file.readlines()
        lines=len(content)
        words=sum(len(line.split()) for line in content)
        chars=sum(len(line) for line in content)
        return lines,words,chars
    
#main 
fileName="data.txt"
l,w,c=countFile(fileName)
print(f"Line counts ={l}\nword counts = {w}\nchracter counts = {c}")
```

```python
'''Write a function that reads a JSON file named `data.json` and prints its contents as a Python dictionary.'''
import json
import os
def readjson(filepath)->dict:
   if os.path.exists(filepath): 
    with open(filepath, 'r') as jf:
        dict = json.load(jf)
   return dict    

#main

#create json file
fileName="data.json"

d={"name":"jalal",
   "age":43,
   "f":"engineer"
   }
with open (fileName,'w') as jf:
    json.dump(d,jf,indent=2)
    
#call readjson to read a json file to python dictionary

json_dict=readjson(fileName)
print(type(json_dict))

# print results

for key,value in  json_dict.items():
    print(f"{key}:{value}")
    
```
