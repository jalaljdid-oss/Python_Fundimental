```python
# by Jalal Jdid 07/02/2026
import random
class Game:
    
    choices = ["rock", "paper", "scissors"]
    user_choice ="None"
    computer_choice = "None"
    user_score = 0
    computer_score = 0
    
    
    def __init__(self,rounds=3):
        self.rounds=rounds
        
    def get_user_choice(self):
        while True:
            try:
                _player_choice=int(input(f"please input one of these choices Only! {self.choices} as (0, 1, 2) in order"))
                if 0 <= _player_choice <= 2:
                    self.user_choice=self.choices[_player_choice]
                    break
            except ValueError:
                print("your choice is wrong please re input only values just  0 or 1 or 2  !!")
            
    
    def get_computer_choice(self):
       #
       self.computer_choice=random.choice(self.choices)
       
        
    
    def determine_winner(self):
        if self.user_choice == self.computer_choice: # same choice it is a tie
           
            print(f"Computer Choose : {self.computer_choice}\nIt is a Tie!\n")
            
        elif self.user_choice == self.choices[0]: # player is rock 
            
            if self.computer_choice == self.choices[2]: # computer is scissors
               
                self.user_score +=1
                print(f"Computer Choose : {self.computer_choice}\n You win this Round!")
                print(f"Current score : You {self.user_score} - {self.computer_score} Computer\n")
                
            else: # computer choice is paper and paper beats rock
              
                self.computer_score +=1
                print(f"Computer Choose : {self.computer_choice}\n Computer wins this Round!")
                print(f"Current score : You {self.user_score} - {self.computer_score} Computer\n")
                
        elif self.user_choice == self.choices[1]: # player is paper
            
            if self.computer_choice == self.choices[0]: # computer is Rock  player wins
               
                self.user_score +=1
                print(f"Computer Choose : {self.computer_choice}\nYou win this Round!")
                print(f"Current score : You {self.user_score} - {self.computer_score} Computer\n")
                
            else: # computer choice is scissors and scissors  beats paper
                
                self.computer_score +=1
                print(f"Computer Choose : {self.computer_choice}\nComputer wins this Round!")
                print(f"Current score : You {self.user_score} - {self.computer_score} Computer\n")
                
        else:  # player is scissors
            
            if self.computer_choice == self.choices[0]: # computer is Rock  computer  wins
                
                self.computer_score +=1
                print(f"Computer Choose : {self.computer_choice}\nComputer wins this Round!")
                print(f"Current score : You {self.user_score} - {self.computer_score} Computer\n")
                
            else: # computer choice is scissors and scissors  beats paper
                
                self.user_score +=1
                print(f"Computer Choose : {self.computer_choice}\nYou win this Round!")
                print(f"Current score : You {self.user_score} - {self.computer_score} Computer\n")
        
            
            
    def play_round(self):
        self.get_user_choice()
        self.get_computer_choice()
        self.determine_winner()
    
    def display_final_result(self):
        print(f"Final score : You {self.user_score} - {self.computer_score} Computer\n")
        if self.user_score > self.computer_score:
            print(f"Congratulation You win The Game!")
        elif self.user_score < self.computer_score:
            print(f"You lose The Game!")
        else:
            print(f"It is a Tie")

    
        
        

# main Program
rounds=int(input("please input the number of Rounds minmum Round is 3 you want to play!"))
if rounds>3:
    my_game=Game(rounds)
    print(f"Welcome to Rock, Paper, Scissors! We'll play {rounds} rounds. Best of luck!")
else:
    my_game=Game()
    print(f"Welcome to Rock, Paper, Scissors! We'll play {my_game.rounds} rounds. Best of luck!")
    


for i in range(my_game.rounds):
    print(f"Round {i+1}:")
    my_game.play_round()
print("\n")    
my_game.display_final_result()   
```
